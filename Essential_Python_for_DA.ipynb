```python
class Dog:                                    # คล้ายๆ cookies cutter
    def __init__(self, name,species, age):    #ข้างหลัง self คือ attribute ที่เราตั้ง   -> ":"ใช้บ่งบอกว่า เป็นจุดเริ่มต้นของชุดคำสั่ง
        self.name = name
        self.species = species
        self.age = age
    def bark(self):
        print(f"{self.name} is barking 'hong hong'!")
    def hbd(self):
        self.age +=1
```

```python
# attributes
dog1 = Dog("milo", "Chihuahua", 2)
dog2 = Dog("david", "Husky", 3)
```

```python
class ATM:
    def __init__(self, ID, balance):
        self.ID = ID
        self.balance = balance

    def __str__(self):
        return f"ID: {self.ID}, Balance: {self.balance}"                   #กรณีเรา print object นี้มันสามารเรียก object นี้ขึ้นมาได้เลย
    def deposit(self, amount):
        self.balance += amount
        print(f"{self.ID}'s balance is {self.balance}")
    def withdraw(self, amount):
        self.balance -= amount
        print(f"{self.ID}'s balance is {self.balance}")
    def check_balance(self):
        print(f"{self.ID}'s balance is {self.balance}")
    def transfer(self, amount, target):
        self.balance -= amount
        target.balance += amount
        print(f"{self.ID}'s balance is {self.balance}")
        print(f"{target.ID}'s balance is {target.balance}")
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
## extension
class Employee(Person):
    def __init__(self,name,age,company):
        super().__init__(name,age)
        self.company = company
    def greeting(self):
        print(f"Hi, my name is {self.name}")
```

```python
Person1 = Person("toy",37)
Employee1= Employee("toy",37,"google")
```

```python
Employee1.greeting()
```

```python
# conext manager -> เปิดปิดไฟลเอง ใช้ with
read_data = []
with open("hotel.csv","r") as file:
    data = csv.reader(file)
    for row in data:
        read_data.append(row)
```

```python
# for loop to cal avg price/night
price = []

for row in read_data[1:]:
    price.append(int(row[4])) ## or -1 for last item of row
print(f"average price is {sum(price)/len(price)}")
```

```python
#Loop for request many charactor
import requests
import time

name=[]
for i in range (10):
    url = f"https://swapi.info/api/people/{i+1}" #เพราะ url เริ่มจากตัวเลข 1
    response = requests.get(url)         # return เป็น response object
    response_json = response.json()      # return เป็น json
    name.append(response_json["name"])
    print(f"Success : {i+1} - {name[i]}")
    time.sleep(1)

print(name)

```
